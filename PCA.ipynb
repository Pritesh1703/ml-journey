```python
'''
Building PCA Algorithm from Scratch without using any library
PCA- Principal component analysis

- Using here for dimensionality reduction
- Create your own data using some mean and variance(optionla.You can use existing data if you have)
-Plot and visualise your data
- Centre the mean(standard scaling using unit variance and zero mean)
- Find covariance matrix
- Find eigen value and eigen vector
- If you want to reduce n dimensions to 2d, then chose top 2 eigen vectors corresponding to top 2 eigen Values which will be your principal component
- Now find the projections of all datapoints on the 2 principal components using dot product
- Now you get data with reduced features without losing much information
'''
```

```python
import numpy as np
import pandas as pd

mu1=np.array([0,0,0])
cov1=np.array([[1,0,0],[0,1,0],[0,0,1]])
sample1= np.random.multivariate_normal(mu1,cov1,20)

df1=pd.DataFrame(sample1,columns=['feature1','feature2','feature3'])
df1['target']=1

mu2=np.array([1,1,1])
cov2=np.array([[1,0,0],[0,1,0],[0,0,1]])
sample2=np.random.multivariate_normal(mu2,cov2,20)

df2=pd.DataFrame(sample2,columns=['feature1','feature2','feature3'])
df2['target']=0

df1= pd.concat([df1,df2],ignore_index=True)
```

```python
df1.head()
```

```python

import plotly.express as px
fig = px.scatter_3d(df1 ,x=df1['feature1'], y=df1['feature2'], z=df1['feature3'],
              color=df1['target'].astype('str'))
fig.update_traces(marker=dict(size=12,
                              line=dict(width=2,
                                        color='DarkSlateGrey')),
                  selector=dict(mode='markers'))

fig.show()
```

```python
# Standard Scaling- Unit variance, zero mean

from sklearn.preprocessing import StandardScaler

scaler=StandardScaler()

df1.iloc[:,0:3] = scaler.fit_transform(df1.iloc[:,0:3])
```

```python
# find a covariance matrix

cov_matrix= np.cov([df1.iloc[:,0],df1.iloc[:,1],df1.iloc[:,2]])
print(cov_matrix)
```

```python
# find eigen value and eigen vector

eigen_values, eigen_vectors=np.linalg.eig(cov_matrix)
```

```python
print("eigen_values",eigen_values)
```

```python

print("eigen_vectors",eigen_vectors)
```

```python
#lets take 2 principal components

pc=eigen_vectors[0:2]
```

```python
# Now lets project all the original data points on these 2 principal components- data.transpose(principal component)

projections = np.dot(df1.iloc[:,0:3],pc.T)
transformed_df= pd.DataFrame(projections,columns=['PC1','PC2'])
transformed_df['target']=df1['target'].values
transformed_df.head()
```

```python

transformed_df['target'] = transformed_df['target'].astype('str')
fig = px.scatter(x=transformed_df['PC1'],
                 y=transformed_df['PC2'],
                 color=transformed_df['target'],
                 color_discrete_sequence=px.colors.qualitative.G10
                )

fig.update_traces(marker=dict(size=12,
                              line=dict(width=2,
                                        color='DarkSlateGrey')),
                  selector=dict(mode='markers'))
fig.show()
```
